**Moving Average Slopes**

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress

# Load data
df = pd.read_csv("USDJPY_data.csv", parse_dates=True, index_col=0)

# Technical indicators
df["Average"] = df[["Open", "High", "Low", "Close"]].mean(axis=1)
df["MA50"] = df["Close"].rolling(50).mean()
df["MA100"] = df["Close"].rolling(100).mean()
df["MA200"] = df["Close"].rolling(200).mean()

# RSI
delta = df["Close"].diff()
gain = delta.clip(lower=0).rolling(14).mean()
loss = (-delta.clip(upper=0)).rolling(14).mean()
df["RSI"] = 100 - (100 / (1 + gain / loss))

# Calculate rolling slopes
def slope(x):
    return linregress(np.arange(len(x)), x)[0]

for col in ["MA50", "MA100", "MA200", "Average", "RSI"]:
    df[f"{col}_Slope"] = df[col].rolling(6).apply(slope, raw=True)

# Plot moving-average slopes
df[["MA50_Slope", "MA100_Slope", "MA200_Slope"]].plot(
    figsize=(12, 5), title="USD/JPY Moving Average Slopes"
)
# Use modern high-contrast styling
plt.style.use("dark_background")

plt.axhline(0, linestyle="--")
plt.xlabel("Date")
plt.ylabel("Slope")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.dates as mdates
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.stats import linregress

# Load data (assuming USDJPY_data.csv is in the working directory)
df = pd.read_csv("USDJPY_data.csv", parse_dates=True, index_col=0)


# 1. DEFINE SLOPE CALCULATION FUNCTION

def get_slope(array):
    y = np.array(array)
    x = np.arange(len(y))
    slope = linregress(x, y)[0]
    return slope



# 2. CALCULATE BASE INDICATORS (Average Price, MA50, MA100, MA200, RSI)

df["Average"] = (df["Open"] + df["High"] + df["Low"] + df["Close"]) / 4
df["MA50"] = df["Close"].rolling(window=50).mean()
df["MA100"] = df["Close"].rolling(window=100).mean()
df["MA200"] = df["Close"].rolling(window=200).mean()

# Relative Strength Index (RSI - 14 Days)
delta = df["Close"].diff()
gain = (delta.where(delta > 0, 0)).rolling(window=14).mean()
loss = (-delta.where(delta < 0, 0)).rolling(window=14).mean()
rs = gain / loss
df["RSI"] = 100 - (100 / (1 + rs))


# 3. GET SLOPE FEATURES OVER ROLLING WINDOW

back_rolling_n = 6

df["slopeMA50"] = (
    df["MA50"].rolling(window=back_rolling_n).apply(get_slope, raw=True)
)
df["slopeMA100"] = (
    df["MA100"].rolling(window=back_rolling_n).apply(get_slope, raw=True)
)
df["slopeMA200"] = (
    df["MA200"].rolling(window=back_rolling_n).apply(get_slope, raw=True)
)
df["AverageSlope"] = (
    df["Average"].rolling(window=back_rolling_n).apply(get_slope, raw=True)
)
df["RSISlope"] = (
    df["RSI"].rolling(window=back_rolling_n).apply(get_slope, raw=True)
)



# 4. HIGH-CONTRAST SHARP PLOT FOR EASY READING

# Use modern high-contrast styling
plt.style.use("dark_background")

fig, ax = plt.subplots(figsize=(14, 5), dpi=300)

# Plot lines with solid/dashed crisp styles and higher line width (lw)
ax.plot(
    df.index,
    df.slopeMA50,
    lw=1.8,
    ls="-",
    color="#FFD700",
    label="slopeMA50 (Fast)",
)
ax.plot(
    df.index,
    df.slopeMA100,
    lw=1.8,
    ls="-",
    color="#00E676",
    label="slopeMA100 (Medium)",
)
ax.plot(
    df.index,
    df.slopeMA200,
    lw=2.0,
    ls="-",
    color="#FF5252",
    label="slopeMA200 (Slow)",
)

# Zero baseline marker
ax.axhline(
    0,
    ls="--",
    lw=1.2,
    color="#FFFFFF",
    alpha=0.7,
    label="Neutral Baseline (0)",
)

# Formatting axes & grid for high readability
ax.set_title(
    "USD/JPY Moving Average Slopes (Trend Momentum)",
    fontsize=14,
    pad=12,
    fontweight="bold",
    color="white",
)
ax.set_xlabel("Year", fontsize=11, fontweight="bold", labelpad=8)
ax.set_ylabel("Slope Angle / Rate of Change", fontsize=11, fontweight="bold")

ax.grid(True, linestyle=":", alpha=0.35, color="gray")
ax.xaxis.set_major_locator(mdates.YearLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))

# Legible Legend
ax.legend(
    loc="upper left",
    frameon=True,
    facecolor="#1E1E1E",
    edgecolor="gray",
    fontsize=10,
)

plt.tight_layout()
plt.show()
```

- **Slope moving toward 0:** Momentum is weakening.

- **All MA slopes > 0:** Strong bullish alignment and possible trend continuation.

- **All MA slopes < 0:** Strong bearish alignment and possible downtrend continuation.

  **CORE TECHNICAL INDICATORS**
 - Average Price (Typical Price)
 - Simple Moving Average (SMA)
 - Relative Strength Index (RSI - 14 Period)
 - Average True Range (ATR - 14 Period)

```python
# CORE TECHNICAL INDICATORS

# Average Price (Typical Price)
df['Average'] = (df['Open'] + df['High'] + df['Low'] + df['Close']) / 4

# Simple Moving Averages (SMA)
df['MA50'] = df['Average'].rolling(window=50).mean()
df['MA100'] = df['Average'].rolling(window=100).mean()
df['MA200'] = df['Average'].rolling(window=200).mean()

# Relative Strength Index (RSI - 14 Period)
delta = df['Average'].diff()
gain = delta.clip(lower=0)
loss = -delta.clip(upper=0)
avg_gain = gain.rolling(window=14).mean()
avg_loss = loss.rolling(window=14).mean()
rs = avg_gain / avg_loss
df['RSI'] = 100 - (100 / (1 + rs))

# Average True Range (ATR - 14 Period)
previous_close = df['Close'].shift(1)
tr1 = df['High'] - df['Low']
tr2 = (df['High'] - previous_close).abs()
tr3 = (df['Low'] - previous_close).abs()
true_range = pd.concat([tr1, tr2, tr3], axis=1).max(axis=1)
df['ATR'] = true_range.rolling(window=14).mean()
```

- **Moving Average:  Distances / Slopes**

```python
# Based on trend prediction, I included the slope of MA50, MA100 and MA200.
# Relative MA Distances (Percentage offset from Moving Averages)
df['Dist_MA50'] = (df['Average'] - df['MA50']) / df['MA50']
df['Dist_MA100'] = (df['Average'] - df['MA100']) / df['MA100']
df['Dist_MA200'] = (df['Average'] - df['MA200']) / df['MA200']

# Moving Average Slopes (5-day Rate of Change)
df['slopeMA50'] = df['MA50'].diff(5)
df['slopeMA100'] = df['MA100'].diff(5)
df['slopeMA200'] = df['MA200'].diff(5)
```

- predict the next-day USD/JPY market trend.

```python
# TARGET: NEXT-DAY BULLISH / BEARISH

# Get the next trading day's Average price
df['Next_Close'] = df['Average'].shift(-1)

# Create target only when a next-day price exists
df['Target'] = np.where(
    df['Next_Close'].notna(),
    (df['Next_Close'] > df['Average']).astype(int),
    np.nan
)

print("\nLAST 5 TARGET OBSERVATIONS")
print(
    df[['Average', 'Next_Close', 'Target']].tail()
)
```

# 10. BACKTEST PERFORMANCE 

```python
import numpy as np
import pandas as pd

# 1. Reconstruct backtest_df directly from your test results
# Assumes df_model has 'Average' (price) and 'Prediction' (0.0 or 1.0) for the test set
backtest_df = df_model.loc[X_test.index, ['Average', 'Prediction']].copy()

# 2. Compute Market Returns from the price column
backtest_df['Market_Return'] = backtest_df['Average'].pct_change().fillna(0)

# 3. Strategy Returns (Shift prediction by 1 day if signals are applied on next day's open)
# Signal alignment: 1 = Long, 0 = Cash (Long-Only) or -1 = Short (Long-Short)
signal_long_only = backtest_df['Prediction'].shift(1).fillna(0)
signal_long_short = backtest_df['Prediction'].replace({0.0: -1.0, 1.0: 1.0}).shift(1).fillna(0)

backtest_df['Ret_LongOnly'] = backtest_df['Market_Return'] * signal_long_only
backtest_df['Ret_LongShort'] = backtest_df['Market_Return'] * signal_long_short

# 4. Compute Cumulative Returns (starting equity = 1.0)
backtest_df['Cum_Market'] = (1 + backtest_df['Market_Return']).cumprod()
backtest_df['Cum_LongOnly'] = (1 + backtest_df['Ret_LongOnly']).cumprod()
backtest_df['Cum_LongShort'] = (1 + backtest_df['Ret_LongShort']).cumprod()

# 5. Define Metric Calculation Function
def calculate_metrics(series_returns, cumulative_series):
    total_return = cumulative_series.iloc[-1] - 1.0
    
    # Annualized Metrics (252 trading days/year)
    n_days = max(len(series_returns), 1)
    ann_return = (cumulative_series.iloc[-1] ** (252 / n_days)) - 1
    ann_vol = series_returns.std() * np.sqrt(252)
    sharpe = ann_return / ann_vol if ann_vol != 0 else 0.0
    
    # Drawdown
    peak = cumulative_series.cummax()
    drawdown = (cumulative_series - peak) / peak
    max_drawdown = drawdown.min()
    
    # Win Rate on active days
    active_days = series_returns[series_returns != 0]
    win_rate = (active_days > 0).mean() if len(active_days) > 0 else 0.0
    
    return {
        "Total Return": f"{total_return * 100:.2f}%",
        "Annualized Return": f"{ann_return * 100:.2f}%",
        "Annualized Volatility": f"{ann_vol * 100:.2f}%",
        "Sharpe Ratio": f"{sharpe:.2f}",
        "Max Drawdown": f"{max_drawdown * 100:.2f}%",
        "Win Rate": f"{win_rate * 100:.2f}%"
    }

# 6. Aggregate Metrics Table
metrics_summary = pd.DataFrame({
    'Buy & Hold': calculate_metrics(backtest_df['Market_Return'], backtest_df['Cum_Market']),
    'Strategy (Long-Only)': calculate_metrics(backtest_df['Ret_LongOnly'], backtest_df['Cum_LongOnly']),
    'Strategy (Long-Short)': calculate_metrics(backtest_df['Ret_LongShort'], backtest_df['Cum_LongShort']),
})

# Display Output
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

print("\n" + "=" * 70)
print("                   BACKTEST PERFORMANCE SUMMARY               ")
print("=" * 70)
print(metrics_summary.to_string())
print("=" * 70)
```
